# 00 张量与计算图

## 目标
学完你能口述：一张 MNIST 图为什么是 `[1,28,28]`；`backward` 为什么需要 `requires_grad`。

## 会错的直觉
「张量就是 numpy，和梯度无关。」——一旦要 `backward`，就必须在计算图上。

### 形状
- 一张灰度图：`[C,H,W] = [1,28,28]`
- 一个 batch：`[B,1,28,28]`
- 十分类 logits：`[B,10]`

```python
import torch
x = torch.zeros(4, 1, 28, 28)
print('batch 图像', tuple(x.shape))
assert x.shape == (4, 1, 28, 28)
w = torch.randn(8, 1, 3, 3, requires_grad=True)
y = torch.nn.functional.conv2d(x, w, padding=1)
print('卷积输出', tuple(y.shape))
assert y.shape == (4, 8, 28, 28)
```

课内练习：下面这格必须成功。若你改了 padding，先想清楚输出高宽。

```python
x = torch.randn(2, 1, 28, 28, requires_grad=True)
loss = x.pow(2).mean()
loss.backward()
assert x.grad is not None
print('grad 均值', float(x.grad.abs().mean()))
```
