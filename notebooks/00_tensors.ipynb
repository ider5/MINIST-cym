# 00 张量与计算图

机器学习把数据表示成**张量**（多维数组）。MNIST 一张图是 `[1, 28, 28]`：通道、高、宽。批量训练时再加 batch 维变成 `[B, 1, 28, 28]`。

PyTorch 用计算图记录运算，`loss.backward()` 沿图反向传播梯度。

```python
import torch
x = torch.randn(2, 1, 28, 28, requires_grad=True)
w = torch.randn(1, 1, 3, 3, requires_grad=True)
y = torch.nn.functional.conv2d(x, w, padding=1).mean()
y.backward()
print('x 的形状', x.shape)
print('w.grad 是否有限', torch.isfinite(w.grad).all().item())
```

练习：把 `requires_grad=True` 去掉再 backward，观察报错。这就是「没有计算图」。
