# 06 超参数实验

## 目标
一次只改一个量；会读 `runs/<model>_<时间戳>/history.json`。

## 会错的直觉
「同时改 lr、batch、模型，谁涨了就是谁的功劳。」——你分不清。

```python
from mnist_lab.data import make_toy_loaders
from mnist_lab.models import build_model
from mnist_lab.train import train_model
from mnist_lab.run_dirs import new_run_dir
tr, va, _ = make_toy_loaders(n=64, batch_size=16)
for lr in (1e-1, 1e-3, 1e-5):
    rd = new_run_dir(f'mlp_lr{lr}', root='runs', stamp=str(lr).replace('.', 'p'))
    r = train_model(build_model('mlp'), tr, va, epochs=2, lr=lr, run_dir=rd)
    print('lr', lr, 'train', r.history['train_loss'], 'val_acc', r.history['val_acc'])
```

课内练习：确认三次实验写进了三个不同目录。

```python
from pathlib import Path
dirs = list(Path('runs').glob('mlp_lr*'))
print([p.name for p in dirs])
assert len(dirs) >= 3
```

读一份 history.json，确认字段。

```python
import json
hist = json.loads(next(Path('runs').glob('mlp_lr*/history.json')).read_text())
print(hist['history'].keys())
assert 'val_acc' in hist['history']
```
