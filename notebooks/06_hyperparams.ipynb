# 06 超参数实验

一次只改一个量：学习率、batch size、epoch、模型。学习率太大损失会炸；太小则学不动。结果写在 `runs/*/history.json`，实验室「超参对照」页可叠加曲线。

```python
from mnist_lab.data import make_toy_loaders
from mnist_lab.models import build_model
from mnist_lab.train import train_model
tr, va, _ = make_toy_loaders(n=64, batch_size=16)
for lr in (1e-1, 1e-3, 1e-5):
    r = train_model(build_model('mlp'), tr, va, epochs=1, lr=lr, run_dir=f'runs/nb06_{lr}', max_steps=4)
    print('lr', lr, 'train_loss', r.history['train_loss'], 'val_acc', r.history['val_acc'])
```
