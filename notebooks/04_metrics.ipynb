# 04 评估指标

## 目标
能手算 3 类混淆矩阵的 precision / recall；知道准确率会掩盖「总把 4 看成 9」。

## 会错的直觉
「准确率 90% 就很好。」——若 90% 样本都是 1，全猜 1 也有 90%。

```python
import numpy as np
from mnist_lab.evaluate import accuracy, confusion_matrix, per_class_report
y_true = np.array([0, 0, 1, 1, 2, 2])
y_pred = np.array([0, 1, 1, 1, 2, 0])
cm = confusion_matrix(y_true, y_pred, num_classes=3)
print(cm)
assert cm[0, 0] == 1 and cm[0, 1] == 1
print('acc', accuracy(y_true, y_pred))
print(per_class_report(cm)['1'])
```

课内填空：类别 0 的 recall 是多少？用断言钉死。

```python
r0 = per_class_report(cm)['0']['recall']
print('class0 recall', r0)
assert abs(r0 - 0.5) < 1e-9
```

再算类别 2 的 precision（预测为 2 的里面有多少真是 2）。

```python
p2 = per_class_report(cm)['2']['precision']
print('class2 precision', p2)
assert abs(p2 - 1.0) < 1e-9
```

准确率手算：6 个里对了几个？

```python
acc = accuracy(y_true, y_pred)
print(acc)
assert abs(acc - 4 / 6) < 1e-9
```
