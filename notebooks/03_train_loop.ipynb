# 03 训练循环与反向传播

## 目标
默写五步：zero_grad → 前向 → loss → backward → step。分清 batch 与 epoch。

## 会错的直觉
「忘了 zero_grad 也没关系，梯度会自动覆盖。」——默认是累加。

```python
from mnist_lab.data import make_toy_loaders
from mnist_lab.models import build_model
from mnist_lab.train import train_model
from mnist_lab.evaluate import evaluate
tr, va, te = make_toy_loaders(n=64, batch_size=16)
model = build_model('mlp')
before = evaluate(model, tr).loss
result = train_model(model, tr, va, epochs=3, lr=0.05, run_dir='runs/nb03')
after = evaluate(model, tr).loss
print('loss', round(before, 3), '→', round(after, 3), 'val_acc', result.history['val_acc'])
assert after < before
```

对照：故意不 zero_grad（见实验室「诊断台」）。

```python
from mnist_lab.bugs import run_bug_case
payload = run_bug_case('forgot_zero_grad', 'runs/nb03_bug')
print(payload['highlight'])
```

数一数：64 个样本、batch=16，一个 epoch 有几个 step？

```python
n_steps = sum(1 for _ in tr)
print('steps / epoch', n_steps)
assert n_steps >= 1
print('history keys', sorted(result.history))
```
