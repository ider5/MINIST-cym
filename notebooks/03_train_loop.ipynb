# 03 训练循环与反向传播

每个 batch：
1. `optimizer.zero_grad()` 清掉上一步梯度
2. `logits = model(x)` 前向
3. `loss = CrossEntropy(logits, y)`
4. `loss.backward()` 反传
5. `optimizer.step()` 用梯度更新参数

`epoch` 是把训练集过一遍；`batch` 是一次更新用的小堆样本。

```python
from mnist_lab.data import make_toy_loaders
from mnist_lab.models import build_model
from mnist_lab.train import train_model
from mnist_lab.evaluate import evaluate
tr, va, te = make_toy_loaders(n=64, batch_size=16)
model = build_model('cnn')
before = evaluate(model, tr).loss
result = train_model(model, tr, va, epochs=2, lr=0.05, run_dir='runs/nb03', max_steps=8)
print('loss 前/后', round(before, 3), result.history['train_loss'])
print('val_acc', result.history['val_acc'])
```
