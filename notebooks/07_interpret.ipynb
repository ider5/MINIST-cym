# 07 模型可解释性

## 目标
能区分特征图、saliency、Grad-CAM；知道课堂必须看**真数字**而不是条纹。

## 会错的直觉
「热力图亮的地方就是像素本身最亮。」——Grad-CAM 亮的是对**当前类别**有贡献的区域。

```python
from mnist_lab.tiny_data import make_tiny_loaders, DEFAULT_CHECKPOINT
from mnist_lab.models import build_model
from mnist_lab.interpret import feature_maps, saliency, gradcam
import torch
model = build_model('cnn')
model.load_state_dict(torch.load(DEFAULT_CHECKPOINT, map_location='cpu'))
_, _, te = make_tiny_loaders(batch_size=8)
x, y = next(iter(te))
x0 = x[:1]
maps = feature_maps(model, x0)
print({k: tuple(v.shape) for k, v in maps.items()})
cam = gradcam(model, x0)
sal = saliency(model, x0)
assert cam.shape == (28, 28) and sal.shape == (28, 28)
print('label', int(y[0]), 'cam max', float(cam.max()))
```

课内练习：CAM 必须是 28×28 且落在 [0,1]。

```python
assert 0.0 <= float(cam.min()) and float(cam.max()) <= 1.0 + 1e-6
print('CAM 范围合法')
```
