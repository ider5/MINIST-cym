# 07 模型可解释性

黑盒也可以部分打开：
- 特征图：中间层每个通道的激活
- saliency：对输入求梯度
- Grad-CAM：最后一层卷积的加权激活，上采样回 28×28

```python
import torch
from mnist_lab.models import build_model
from mnist_lab.interpret import feature_maps, saliency, gradcam
from mnist_lab.data import make_toy_loaders
from mnist_lab.train import train_model
tr, va, te = make_toy_loaders(n=48, batch_size=16)
model = build_model('cnn')
train_model(model, tr, va, epochs=1, lr=0.05, run_dir='runs/nb07', max_steps=6)
x, y = next(iter(te))
x0 = x[:1]
print({k: v.shape for k, v in feature_maps(model, x0).items()})
print('saliency', saliency(model, x0).shape, 'gradcam', gradcam(model, x0).shape)
```
