# 01 数据管道、划分与增强

## 目标
能解释 train / val / test 各自只能干什么；知道增强发生在训练期。

## 会错的直觉
「准确率最高的那个划分方式最好。」——如果拿测试集反复试，数字会虚高。

```python
from mnist_lab.data import make_toy_loaders, split_train_val, mnist_transforms
from torch.utils.data import TensorDataset
import torch
train, val, test = make_toy_loaders(n=80, batch_size=16)
print(len(train.dataset), len(val.dataset), len(test.dataset))
bx, by = next(iter(train))
assert bx.shape[1:] == (1, 28, 28)
print('一个 batch', tuple(bx.shape), by[:4].tolist())
```

真实数字请用仓库里的 `fixtures/mnist_tiny.pt`（条纹 toy 只适合测 API）：

```python
from mnist_lab.tiny_data import make_tiny_loaders
tr, va, te = make_tiny_loaders(batch_size=16)
x, y = next(iter(tr))
print('tiny batch', tuple(x.shape), 'labels', y[:8].tolist())
assert x.min() >= 0 and x.max() <= 1
```

```python
print(mnist_transforms(augment=False))
print(mnist_transforms(augment=True))
```

核对划分比例（100 条、val_ratio=0.2 → 80/20）。

```python
ds = TensorDataset(torch.zeros(100, 1, 28, 28), torch.zeros(100, dtype=torch.long))
a, b = split_train_val(ds, val_ratio=0.2, seed=0)
print(len(a), len(b))
assert len(a) == 80 and len(b) == 20
```
